# NumPy — Complete Beginner-to-Intermediate Notes

A practical notebook covering the core NumPy concepts used in Python programming, Data Analytics, Data Science and Machine Learning.



## 1. What is NumPy?

**NumPy (Numerical Python)** is a Python library used for fast numerical computing. Its main object is the `ndarray` (N-dimensional array).

### Why use NumPy?
- Stores numerical data efficiently.
- Performs operations on entire arrays without writing Python loops.
- Supports multidimensional arrays and mathematical functions.
- Provides tools for statistics, random numbers and linear algebra.

### NumPy vs. Python lists
Python lists can store mixed data types. NumPy arrays are generally designed for elements of a common data type and are optimized for numerical operations.

```python
import numpy as np

print("NumPy version:", np.__version__)

python_list = [1, 2, 3]
numpy_array = np.array([1, 2, 3])

print("Python list:", python_list)
print("NumPy array:", numpy_array)
print("List multiplied by 2:", python_list * 2)
print("Array multiplied by 2:", numpy_array * 2)
```

## 2. Creating NumPy arrays

The `np.array()` function converts a list or tuple into an array.

Common creation functions:
- `np.array()` — create from existing data.
- `np.zeros()` — create an array filled with zeros.
- `np.ones()` — create an array filled with ones.
- `np.full()` — fill an array with a chosen value.
- `np.arange()` — create values using a step.
- `np.linspace()` — create a fixed number of evenly spaced values.
- `np.eye()` — create an identity matrix.
- `np.empty()` — allocate an array without initializing its values.

```python
a = np.array([10, 20, 30, 40])
b = np.array([[1, 2, 3], [4, 5, 6]])

print("1D array:", a)
print("2D array:\n", b)
print("Zeros:\n", np.zeros((2, 3)))
print("Ones:", np.ones(4))
print("Full:", np.full((2, 2), 7))
print("Arange:", np.arange(0, 10, 2))
print("Linspace:", np.linspace(0, 1, 5))
print("Identity matrix:\n", np.eye(3))
```

## 3. Dimensions, shape, size and data type

Important array attributes:
- `ndim` — number of dimensions.
- `shape` — length along each dimension.
- `size` — total number of elements.
- `dtype` — element data type.
- `itemsize` — bytes used by one element.
- `nbytes` — total bytes used by the array elements.

```python
arr = np.array([[1, 2, 3], [4, 5, 6]], dtype=np.int32)

print("Array:\n", arr)
print("Dimensions:", arr.ndim)
print("Shape:", arr.shape)
print("Total elements:", arr.size)
print("Data type:", arr.dtype)
print("Bytes per element:", arr.itemsize)
print("Total element storage in bytes:", arr.nbytes)
```

## 4. Data types and type conversion

NumPy supports integer, floating-point, Boolean, complex and other data types.

Use `dtype` when creating an array, or `.astype()` to convert an existing array. Conversion can lose information, such as decimal places when converting floats to integers.

```python
integers = np.array([1, 2, 3], dtype=np.int64)
decimals = np.array([1.5, 2.7, 3.9])
flags = np.array([True, False, True])

print(integers.dtype, decimals.dtype, flags.dtype)

converted = decimals.astype(int)
print("Converted to integers:", converted)
```

## 5. Indexing and accessing elements

Indexing starts at **0**. For a 2D array, use `array[row, column]`.

```python
arr = np.array([[10, 20, 30], [40, 50, 60], [70, 80, 90]])

print("First element:", arr[0, 0])
print("Second row:", arr[1])
print("Third column:", arr[:, 2])
print("Last element:", arr[-1, -1])
```

## 6. Slicing arrays

Basic syntax: `array[start:stop:step]`. The stop position is excluded.

For 2D arrays, slice rows and columns separately: `array[row_slice, column_slice]`.

```python
numbers = np.array([0, 10, 20, 30, 40, 50, 60])

print(numbers[1:5])
print(numbers[:4])
print(numbers[3:])
print(numbers[::2])
print(numbers[::-1])

matrix = np.arange(1, 13).reshape(3, 4)
print("Original matrix:\n", matrix)
print("First two rows, columns 1 to 2:\n", matrix[:2, 1:3])
```

## 7. Boolean indexing and filtering

A condition creates a Boolean array. Use it to select values that meet a requirement.

```python
marks = np.array([45, 82, 67, 91, 38, 76])

passed = marks >= 40
print("Condition:", passed)
print("Passing marks:", marks[passed])
print("Marks above 70:", marks[marks > 70])

# Combine conditions with & (and), | (or), and ~ (not).
selected = marks[(marks >= 60) & (marks <= 85)]
print("Marks from 60 to 85:", selected)
```

## 8. Fancy indexing

Fancy indexing selects elements using a list or array of integer positions.

```python
values = np.array([100, 200, 300, 400, 500])
positions = [0, 2, 4]

print(values[positions])

matrix = np.arange(1, 13).reshape(3, 4)
print("Selected rows:\n", matrix[[0, 2]])
print("Selected elements:", matrix[[0, 1, 2], [1, 2, 3]])
```

## 9. Changing array shape

Useful methods:
- `reshape()` — change dimensions without changing the element count.
- `flatten()` — return a flattened copy.
- `ravel()` — return a flattened view when possible.
- `resize()` — change the array's shape in place.
- `T` or `transpose()` — swap axes.
- `squeeze()` — remove dimensions of length 1.
- `expand_dims()` — add a dimension.

```python
arr = np.arange(1, 13)
matrix = arr.reshape(3, 4)

print("3 x 4:\n", matrix)
print("2 x 6:\n", arr.reshape(2, 6))
print("Flatten:", matrix.flatten())
print("Ravel:", matrix.ravel())
print("Transpose:\n", matrix.T)

single_dim = np.array([[[5, 6, 7]]])
print("Squeezed:", np.squeeze(single_dim))
print("Expanded shape:", np.expand_dims(np.array([1, 2, 3]), axis=0).shape)
```

## 10. Joining and splitting arrays

- `concatenate()` joins along an existing axis.
- `stack()` joins arrays along a new axis.
- `vstack()` stacks vertically.
- `hstack()` stacks horizontally.
- `split()` divides an array into equal sections where possible.
- `array_split()` allows sections of unequal sizes.

```python
x = np.array([1, 2, 3])
y = np.array([4, 5, 6])

print("Concatenate:", np.concatenate((x, y)))
print("Stack:\n", np.stack((x, y)))
print("Vertical stack:\n", np.vstack((x, y)))
print("Horizontal stack:", np.hstack((x, y)))

parts = np.array_split(np.arange(10), 3)
print("Split parts:", parts)
```

## 11. Arithmetic operations

NumPy arithmetic is usually element-wise. The arrays must have compatible shapes.

```python
a = np.array([10, 20, 30])
b = np.array([1, 2, 3])

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Power:", a ** 2)
print("Remainder:", a % 7)
print("Comparison:", a > 15)
```

## 12. Universal functions (ufuncs)

Universal functions perform operations element-wise efficiently.

Examples include `sqrt`, `square`, `exp`, `log`, `sin`, `cos`, `abs`, `round`, `floor` and `ceil`.

```python
values = np.array([1, 4, 9, 16])

print("Square root:", np.sqrt(values))
print("Square:", np.square(values))
print("Exponential:", np.exp(values))
print("Natural log:", np.log(values))
print("Absolute values:", np.abs(np.array([-2, -1, 0, 1, 2])))
print("Rounded:", np.round(np.array([1.234, 5.678]), 2))
```

## 13. Aggregation and statistical functions

Common functions:
- `sum()`, `min()`, `max()`
- `mean()`, `median()`
- `std()` (standard deviation), `var()` (variance)
- `percentile()`, `quantile()`
- `argmin()`, `argmax()` (positions of minimum and maximum)

For multidimensional arrays, `axis=0` works down rows (column-wise result), while `axis=1` works across columns (row-wise result).

```python
data = np.array([[10, 20, 30], [40, 50, 60]])

print("Sum:", np.sum(data))
print("Minimum:", np.min(data))
print("Maximum:", np.max(data))
print("Mean:", np.mean(data))
print("Median:", np.median(data))
print("Standard deviation:", np.std(data))
print("Variance:", np.var(data))
print("75th percentile:", np.percentile(data, 75))
print("Index of maximum in flattened array:", np.argmax(data))

print("Column sums (axis=0):", np.sum(data, axis=0))
print("Row sums (axis=1):", np.sum(data, axis=1))
```

## 14. Broadcasting

Broadcasting lets NumPy operate on arrays with different but compatible shapes. Dimensions are compatible when they are equal or one of them is 1.

This is especially useful for adding a value to every element, scaling columns, or applying calculations to matrices.

```python
matrix = np.array([[1, 2, 3], [4, 5, 6]])
print("Add 10 to every element:\n", matrix + 10)

column_values = np.array([100, 200, 300])
print("Add values to each column:\n", matrix + column_values)

row_values = np.array([[10], [20]])
print("Add values to each row:\n", matrix + row_values)
```

## 15. Sorting and searching

- `np.sort()` returns a sorted copy.
- `array.sort()` sorts the array in place.
- `argsort()` returns the indices that would sort the array.
- `where()` returns positions where a condition is true, or selects between values.
- `unique()` returns sorted unique values.
- `searchsorted()` finds insertion positions in a sorted array.

```python
values = np.array([40, 10, 30, 20, 10])

print("Sorted copy:", np.sort(values))
print("Sorting indices:", np.argsort(values))
print("Unique values:", np.unique(values))
print("Positions where value > 20:", np.where(values > 20))
print("Replace values below 25:", np.where(values < 25, 0, values))

sorted_values = np.array([10, 20, 30, 40])
print("Insertion position for 25:", np.searchsorted(sorted_values, 25))
```

## 16. Handling missing and special values

NumPy commonly represents missing numeric values with `np.nan`. Note that `NaN` is not equal to itself; use `np.isnan()` to detect it.

Other useful checks include `np.isfinite()`, `np.isinf()` and `np.isfinite()`.

```python
values = np.array([10.0, np.nan, 30.0, np.inf, -np.inf])

print("Is NaN:", np.isnan(values))
print("Is infinite:", np.isinf(values))
print("Is finite:", np.isfinite(values))

valid_values = values[np.isfinite(values)]
print("Finite values:", valid_values)

clean_values = np.nan_to_num(values, nan=0, posinf=999, neginf=-999)
print("Replace special values:", clean_values)
```

## 17. Random number generation

Use `np.random.default_rng()` to create a random-number generator. This is the recommended modern interface.

A seed makes results reproducible.

```python
rng = np.random.default_rng(seed=42)

print("Random decimal:", rng.random())
print("Random integers:", rng.integers(1, 10, size=5))
print("Normal distribution:", rng.normal(loc=0, scale=1, size=5))
print("Random choice:", rng.choice(["red", "blue", "green"], size=4))
print("Shuffled values:", rng.permutation(np.arange(1, 6)))
```

## 18. Linear algebra basics

NumPy's `linalg` module supports common linear algebra operations:
- `dot()` / `matmul()` — matrix multiplication.
- `linalg.det()` — determinant.
- `linalg.inv()` — inverse (when it exists).
- `linalg.eig()` — eigenvalues and eigenvectors.
- `linalg.solve()` — solve a system of linear equations.
- `linalg.norm()` — vector or matrix norm.

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print("Matrix multiplication:\n", A @ B)
print("Dot product:", np.dot(np.array([1, 2]), np.array([3, 4])))
print("Determinant:", np.linalg.det(A))
print("Inverse:\n", np.linalg.inv(A))
print("Norm:", np.linalg.norm(A))

# Solve: 2x + y = 5 and x + 3y = 6
coefficients = np.array([[2, 1], [1, 3]])
answers = np.array([5, 6])
solution = np.linalg.solve(coefficients, answers)
print("Solution [x, y]:", solution)
```

## 19. Copy vs. view

A **view** shares data with the original array. A **copy** owns separate data.

Basic slicing often creates a view; fancy indexing and `.copy()` create copies. Changes to a view can therefore affect the original array.

```python
original = np.array([10, 20, 30, 40])

view = original[1:3]
view[0] = 999
print("Original after changing view:", original)

original = np.array([10, 20, 30, 40])
copy_array = original.copy()
copy_array[0] = 999
print("Original after changing copy:", original)
print("Copy:", copy_array)

print("Shares memory (view):", np.shares_memory(original, view))
```

## 20. Saving and loading arrays

NumPy can save arrays in its own binary format or in text formats such as CSV.

- `np.save()` / `np.load()` — `.npy` format.
- `np.savez()` — store multiple arrays in one `.npz` file.
- `np.savetxt()` / `np.loadtxt()` — text files such as CSV.

```python
sample = np.array([[1, 2, 3], [4, 5, 6]])

np.save("sample_array.npy", sample)
loaded = np.load("sample_array.npy")
print("Loaded NPY file:\n", loaded)

np.savetxt("sample_array.csv", sample, delimiter=",", fmt="%d")
loaded_csv = np.loadtxt("sample_array.csv", delimiter=",")
print("Loaded CSV file:\n", loaded_csv)
```

## 21. Vectorization: replacing loops with array operations

Vectorization means expressing calculations as operations on whole arrays. It often makes code shorter and can improve performance compared with explicit Python loops.

```python
numbers = np.arange(1, 6)

# Vectorized approach
squares = numbers ** 2
print("Squares:", squares)

# Compare with a regular Python loop
loop_squares = []
for number in numbers:
    loop_squares.append(number ** 2)

print("Loop result:", loop_squares)
```

## 22. Practical mini-project: student marks analysis

Use NumPy to analyse marks for five students across three subjects. This combines array creation, shape, aggregation, indexing, filtering and sorting.

```python
student_names = np.array(["Asha", "Riya", "Kabir", "Meera", "Arjun"])
marks = np.array([
    [78, 85, 90],
    [92, 88, 95],
    [56, 70, 65],
    [81, 79, 84],
    [67, 73, 72]
])

student_average = np.mean(marks, axis=1)
subject_average = np.mean(marks, axis=0)
overall_average = np.mean(marks)

print("Student averages:")
for name, average in zip(student_names, student_average):
    print(name, ":", round(average, 2))

print("\nSubject averages:", np.round(subject_average, 2))
print("Overall average:", round(overall_average, 2))

top_student_index = np.argmax(student_average)
print("Top student:", student_names[top_student_index])

passed_students = student_names[student_average >= 70]
print("Students with average >= 70:", passed_students)
```
